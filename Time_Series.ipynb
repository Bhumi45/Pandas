##### Examples:
 -  Financial Data(Company stocks)
 -  Natural Data(Rainfall measurement)
 -  Event Data(Covid)
 -  Medical Data

##### Types of operations done on Time Series

 - Time Series Analysis
 - Time Series Forecasting

## **Resampling**

Resampling involves changing the frequency of your time series observations.

Two types of resampling are:

**Unsampling** : where you can increase the freq of the samples , such as from minutes to seconds.
**Downsampling**: Where you decrease the frequency of the samples, such as from days to months.


Genereally we do downsampling to decrease the noise and undersatand the data better.

```python
google['Close'].plot()
```

```python
google['Close'].plot()
google['Close'].resample('W').mean().plot()
```

```python
google['Volume'].plot()
google['Volume'].resample('W').mean().plot()
```

```python
''' upsampling''' # since km number se jada numbers of rows bn jati hai toh there wuld be nan values ,which we need to fill.
google['Close'].resample('h').interpolate(method='linear').plot()
```

```python
google['Close'].resample('h').interpolate(method='spline',order=2).plot()
```

## **Rolling Window**

Times Series data in original format can be quite volatile, especially on smaller agregation levels.The concept of rolling , or moving averages is a useful technique for smoothing time series data.(it is use to smoothen the data).

There are two types of moving averages:

1.**Rolling average**: Give equal to weightage to all numbers.

2.**Exponential Moving Average**: give higher weightage to the recent number

```python
google['Close'].rolling(3).mean()
```

```python
google['Close'].plot()
google['Close'].rolling(3).mean().plot()
```

```python
google['Volume'].plot()
google['Volume'].rolling(10).mean().plot()
```

```python
google['Volume'].plot(legend='original')
google['Volume'].rolling(10).mean().plot(legend='rolling')
google['Volume'].ewm(10).mean().plot(legend='ewm')
```

```python
google['Volume'].rolling(10).mean().plot(legend='rolling')
google['Volume'].ewm(10).mean().plot(legend='ewm')
```
